```python
import pandas as pd
from ete3 import Tree
```

```python
tree = Tree('/home/pmh3ax/pprs_analysis/rfl-tree/trim_rfl_seqs.fa.contree', format=1)

threshold = 0.1

for node in tree.traverse("postorder"):
    if not node.is_leaf():
        leaves = node.get_leaves()
        max_distance = max([leaf.dist for leaf in leaves])
        if max_distance < threshold:
            representative = max(leaves, key=lambda leaf: leaf.dist)
            for leaf in leaves:
                if leaf != representative:
                    leaf.delete()
                    
unique_seqs = [leaf.name for leaf in tree.get_leaves()]
tree.write(outfile='/home/pmh3ax/pprs_analysis/rfl-tree/collapse_rfl_simplified.nwk')

with open('/home/pmh3ax/pprs_analysis/rfl-tree/collapse_rfl_seqs.txt', "w") as f:
    f.write("\n".join(unique_seqs))
print(f"Retained {len(unique_seqs)} unique sequences for BLAST.")
```
